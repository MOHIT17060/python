```python
import numpy as np
```

```python
#join two array
arr1=np.array([1,2,3,4])
arr2=np.array([1,2,3,4])
arr=np.concatenate((arr1,arr2))
print(arr)
```

```python
arr1=np.array([1,2,3,4])
arr2=np.array([[1,2,3,4]])
arr=np.concatenate((arr1,arr2))
print(arr)
```

```python
arr1=np.array([[1,2,3,4,5]])
arr2=np.array([[6,7,8,9,10]])
print(arr1)
print(arr2)
print()
arr=np.concatenate((arr1,arr2),axis=0)
print(arr)
arrn=np.concatenate((arr1,arr2),axis=1)
print(arrn)
```
